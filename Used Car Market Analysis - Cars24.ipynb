```python
import numpy as np
import pandas as pd
from bs4 import BeautifulSoup
from curl_cffi import requests
import re
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import MaxNLocator
```

## Data Extraction 

```python
brands = ['maruti','hyundai','tata','mahindra', 'honda','toyota',
          'kia','renault','volkswagen', 'skoda','ford','mg',
          'jeep','nissan','audi']

cities = ['delhi-ncr','mumbai','bangalore','hyderabad', 
          'chennai','pune','kolkata','ahmedabad','jaipur', 
          'lucknow','gurgaon','noida','ghaziabad','kochi', 
          'nashik','nagpur','coimbatore','indore','patna',
          'chandigarh', 'surat','visakhapatnam','agra','bhopal','rajkot']

# Standard realistic Chrome headers
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36',
    'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,*/*;q=0.8',
    'Accept-Language': 'en-US,en;q=0.9',
    'Accept-Encoding': 'gzip, deflate, br',
}

car_name = []
brand_name = []
year = []
model_name = []
km_driven = []
transmission_type = []
fuel_type = []
city_name = []
price = [] 
owner_type = []

for brand in brands:
    for city in cities:
        url = f"https://www.cars24.com/buy-used-{brand}-cars-{city}"
        print(f"Target URL: {url}")


        # Impersonate a real Chrome browser instance
        response = requests.get(url, headers=headers, impersonate="chrome120")

        print("Status Code:", response.status_code)
        if response.status_code == 200:
            print("Successfully retrieved page content!")

            soup = BeautifulSoup(response.text, "html.parser")
            
        for a in soup.find_all('span',class_='sc-bcXHqh bAcffq'):
            text = a.get_text('',strip=True)
            car_name.append(text)
        
        for b in soup.find_all('span',class_='sc-bcXHqh bAcffq'):
            text = b.get_text('',strip=True)
            match = re.search(r'\d+\s([A-Za-z]+)',text)
            if match:
                brand_name.append(match.group(1))
            else:
                brand_name.append(None)

        for c in soup.find_all('span',class_='sc-bcXHqh bAcffq'):
            text = c.get_text('',strip=True)
            match = re.search(r'(\d{4})\s',text)
            if match:
                year.append(match.group(1))
            else:
                year.append(None)

        for d in soup.find_all('span',class_='sc-bcXHqh bAcffq'):
            text = d.get_text('',strip=True)
            match = re.search(r'\d{4}\s[A-Za-z]+\s(.+)',text)
            if match:
                model_name.append(match.group(1).strip())
            else:
                model_name.append(None)

        for e in soup.find_all('span', class_='sc-bcXHqh bAcffq'):
            text = e.get_text(' ', strip=True)
            card = e.parent.parent.parent
            card_text = card.get_text(' ', strip=True)
            match = re.search(r'[\d,]+\s?km', card_text)
            if match:
                km_driven.append(match.group(0))
            else:
                km_driven.append(None)

        for f in soup.find_all('span', class_='sc-bcXHqh bAcffq'):
            card = f.parent.parent.parent
            card_text = card.get_text(' ', strip=True)

            if 'Manual' in card_text:
                transmission_type.append('Manual')
            elif 'Auto' in card_text:
                transmission_type.append('Auto')
            else:
                transmission_type.append(None)

        for g in soup.find_all('span', class_='sc-bcXHqh bAcffq'):
            card = g.parent.parent.parent
            card_text = card.get_text(' ', strip=True)

            if 'CNG' in card_text:
                fuel_type.append('CNG')
            elif 'Petrol' in card_text:
                fuel_type.append('Petrol')
            elif 'Diesel' in card_text:
                fuel_type.append('Diesel')
            elif 'Electric' in card_text:
                fuel_type.append('Electric')
            elif 'Hybrid' in card_text:
                fuel_type.append('Hybrid')
            else:
                fuel_type.append(None)

        for h in soup.find_all('span', class_='sc-bcXHqh bAcffq'):
            city_name.append(city)

        for i in soup.find_all('span', class_='sc-bcXHqh bAcffq'):
            card = i.parent.parent.parent.parent

            price_box = card.find('div', class_='styles_priceWrap__VwWBV')

            if price_box:
                p = price_box.find_all('p')

                if len(p)>=2:
                    price.append(p[1].get_text(strip=True))
                else:
                    price.append(None)
            else:
                price.append(None)

        for j in soup.find_all('span', class_='sc-bcXHqh bAcffq'):

            card_link = j.parent.parent.parent.parent.parent.parent.parent
            href = card_link.get('href')
        
            if href and re.search(r'-\d+/$', href):
                car_url = href
        
                response2 = requests.get(
                    car_url,
                    headers=headers,
                    impersonate='chrome120'
                )
                car_soup = BeautifulSoup(response2.text, 'html.parser')
        
                ownership = None
        
                for p in car_soup.find_all('p'):
                    if p.get_text(' ', strip=True) == 'Ownership':
                        parent = p.parent
                        values = parent.find_all('p')
        
                        if len(values) >= 2:
                            ownership = values[1].get_text(' ', strip=True)
                        break
                owner_type.append(ownership)
        
            else:
                owner_type.append(None)
```

```python
print('Brand Name: ',len(brand_name))
print('Year: ',len(year))
print('Model Name: ',len(model_name))
print('Car Name: ',len(car_name))
print('KM Driven: ',len(km_driven))
print('Transmission Type: ',len(transmission_type))
print('Fuel Type: ',len(fuel_type))
print('Cities name: ',len(city_name))
print('Price: ',len(price))
print('Owner Type:',len(owner_type))
```

### Creation of Data Frame

```python
cars = pd.DataFrame({'Car Name':car_name,
    'Brand Name':brand_name,
    'Model Name':model_name,
    'KM Driven':km_driven,
    'Transmission Type':transmission_type,
    'Price':price,
    'Fuel Type':fuel_type,
    'Year':year,
    'Cities name':city_name,
    'Owner Type':owner_type})
```
